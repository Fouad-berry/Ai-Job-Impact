## 3. Distributions numériques

```python
df.select_dtypes(include=[np.number]).hist(figsize=(15, 10), bins=25, edgecolor='black')
plt.suptitle('Distributions des variables numériques', y=1.02, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Variables catégorielles

```python
cat_cols = ['Gender', 'Education_Level', 'Industry', 'AI_Adoption_Level',
            'Automation_Risk', 'Job_Status', 'Upskilling_Required', 'Remote_Work']
fig, axes = plt.subplots(2, 4, figsize=(18, 9))
for ax, col in zip(axes.flatten(), cat_cols):
    df[col].value_counts().plot(kind='bar', ax=ax, color='steelblue', edgecolor='black')
    ax.set_title(col); ax.tick_params(axis='x', rotation=30)
plt.tight_layout(); plt.show()
```

## 5. Variable cible — Job Status

```python
counts = df['Job_Status'].value_counts()
print(counts)
print(f'\n% remplacés : {100*(df["Job_Status"]=="Replaced").mean():.2f}%')
print(f'% modifiés  : {100*(df["Job_Status"]=="Modified").mean():.2f}%')
print(f'% inchangés : {100*(df["Job_Status"]=="Unchanged").mean():.2f}%')
```

## 6. Premier aperçu : salaire avant/après IA

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(df['Salary_Before_AI'], df['Salary_After_AI'], alpha=0.3, s=15)
ax.plot([30000, 130000], [30000, 130000], 'r--', label='y = x (pas de changement)')
ax.set_xlabel('Salaire avant IA'); ax.set_ylabel('Salaire après IA')
ax.set_title('Salaire avant vs après IA'); ax.legend()
plt.show()

print(f'Δ salaire moyen : ${(df["Salary_After_AI"]-df["Salary_Before_AI"]).mean():,.0f}')
```
